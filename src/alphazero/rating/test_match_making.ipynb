# Verify Match Makings Work

Here, the rating system are not run to rate, using pre-defined ratings.

```python
import time

import pandas as pd
import matplotlib.pyplot as plt

from rating import Game, Games
from elo import EloSystem

from match_making import MatchMaking
from match_making_implementations import (
    RatingMatchMaking,
    WinningProbabilityMatchMaking,
)
```

# Simple Scenario

```python
elo_rating = EloSystem()

elo_rating.add_player(name="Player A", rating=1500, games_played=51)
elo_rating.add_player(name="Player B", rating=800, games_played=27)
elo_rating.add_player(name="Player C")
elo_rating.add_player(name="Player D")
```

```python
elo_rating.display_players()
```

### Rating 

```python
rating_linear_match_making = RatingMatchMaking(
    rating_system=elo_rating, optimization_method="linear", verbose=True
)

rating_linear_match_making.match_players()

rating_linear_match_making.matched_pairs
```

```python
_ = rating_linear_match_making.evaluate_match_making_rating_differences()
```

```python
_ = rating_linear_match_making.evaluate_match_making_winning_probability_differences()
```

```python
rating_hungarian_match_making = RatingMatchMaking(
    rating_system=elo_rating, optimization_method="hungarian", verbose=True
)

rating_hungarian_match_making.match_players()

rating_hungarian_match_making.matched_pairs
```

```python
_ = rating_hungarian_match_making.evaluate_match_making_rating_differences()
```

```python
_ = (
    rating_hungarian_match_making.evaluate_match_making_winning_probability_differences()
)
```

### Winning Probability

```python
winning_probability_linear_match_making = WinningProbabilityMatchMaking(
    rating_system=elo_rating, optimization_method="linear", verbose=True
)

winning_probability_linear_match_making.match_players()

winning_probability_linear_match_making.matched_pairs
```

```python
_ = winning_probability_linear_match_making.evaluate_match_making_rating_differences()
```

```python
_ = (
    winning_probability_linear_match_making.evaluate_match_making_winning_probability_differences()
)
```

```python
winning_probability_hungarian_match_making = WinningProbabilityMatchMaking(
    rating_system=elo_rating, optimization_method="hungarian", verbose=True
)

winning_probability_hungarian_match_making.match_players()

winning_probability_hungarian_match_making.matched_pairs
```

```python
_ = (
    winning_probability_hungarian_match_making.evaluate_match_making_rating_differences()
)
```

```python
_ = (
    winning_probability_hungarian_match_making.evaluate_match_making_winning_probability_differences()
)
```

### Comparison

```python
match_making_systems: dict[str, MatchMaking] = {
    "rating - linear": rating_linear_match_making,
    "rating - hungarian": rating_hungarian_match_making,
    "winnings odds -  linear": winning_probability_linear_match_making,
    "winning odds - hungarian": winning_probability_hungarian_match_making,
}

method_to_median_abs_difference: dict[str, float] = {}

plt.figure()
for name, match_making_system in match_making_systems.items():
    verbose_state = match_making_system.verbose
    match_making_system.verbose = False
    difference, abs_difference = (
        match_making_system.evaluate_match_making_rating_differences()
    )
    plt.hist(difference, bins=20, label=name, alpha=0.5)
    match_making_system.verbose = verbose_state

    method_to_median_abs_difference[name] = abs_difference.abs().median()

plt.legend()
plt.title("Different Match Making Systems Comparisons\nDifference in Rating per Pairs")
plt.show()

methods_median_abs_rating_differences = pd.Series(method_to_median_abs_difference)

methods_median_abs_rating_differences
```

```python
match_making_systems: dict[str, MatchMaking] = {
    "rating - linear": rating_linear_match_making,
    "rating - hungarian": rating_hungarian_match_making,
    "winnings odds -  linear": winning_probability_linear_match_making,
    "winning odds - hungarian": winning_probability_hungarian_match_making,
}


method_to_median_abs_difference: dict[str, float] = {}

plt.figure()
for name, match_making_system in match_making_systems.items():
    verbose_state = match_making_system.verbose
    match_making_system.verbose = False
    difference, abs_difference = (
        match_making_system.evaluate_match_making_winning_probability_differences()
    )
    plt.hist(difference, bins=20, label=name, alpha=0.5)
    match_making_system.verbose = verbose_state

    method_to_median_abs_difference[name] = abs_difference.abs().median()

plt.legend()
plt.title(
    "Different Match Making Systems Comparisons\nDifference in Winning Odds per Pairs"
)
plt.show()

methods_median_abs_rating_differences = pd.Series(method_to_median_abs_difference)

methods_median_abs_rating_differences
```
